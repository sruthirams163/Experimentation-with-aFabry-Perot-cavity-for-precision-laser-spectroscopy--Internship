```python
import numpy as np
import matplotlib.pyplot as plt


D= 100*10**(-2)

def intensity(x, y, d, D, wavelength):
    E = E1 + E2
    I = 0.5 * (E**2) * 8.854 * 10**(-12) * (3 * 10**8)**2
    return I

def euclideanDistance(x, y):
    dist = np.sqrt((x[0] - y[0])**2 + (x[1] - y[1])**2)
    return dist
    def find_hypotenuse(dist, D):
        s1p = D**2 + (x**2+ y**2)**0.5
        return s1p
        s1p = find_hypotenuse(dist, D)

    dist1 = euclideanDistance([x, y], [0, -d/2])
    def find_hypotenuse(dist1, D):
        s2p = D**2 + (x**2+ y**2)**0.5
        return s2p
        s2p = find_hypotenuse(dist1, D)

def e1(s1p, wavelength):
    theta = (2 * np.pi * s1p) / wavelength
    return e01 * np.cos(theta)

def e2(s2p, wavelength):
    theta = (2 * np.pi * s2p) / wavelength
    return e02 * np.cos(theta)

wavelength = 550*10**(-9)
e01 = 1
e02 = 1

d = 0.09*10**(-3)

x = np.linspace(-2*10**(-3) , 2*10**(-3)  , 1000)
y = np.linspace(-2*10**(-3)  ,2*10**(-3) , 1000)

[X, Y] = np.meshgrid(x, y)

s1p = euclideanDistance([X, Y], [0, d/2])
s2p = euclideanDistance([X, Y], [0, -d/2])

E1 = e1(s1p, wavelength)
E2 = e2(s2p, wavelength)

I = intensity(X, Y, d, D, wavelength)

plt.figure(figsize=(11, 6))

plt.imshow(I, extent=[np.min(x), np.max(x), np.min(y), np.max(y)], cmap='gray')
plt.colorbar(label='E')
plt.title('Interference Pattern')


plt.show()


 
```
