```python
import numpy as np
import matplotlib.pyplot as plt

def fn_trans(f, freq, linewidth=1):
    delta = freq  / linewidth
    return 1 /(1 +(2*f*delta / np.pi)**2)

freq = np.linspace(-5, 5, 1000) 
linewidth=1
f_values = [2, 10, 50, 100, 2000]

plt.figure(figsize=(10, 6))
for f in f_values:
    trans = fn_trans(f, freq, linewidth)
    plt.plot(freq, trans, label=f'Finesse = {f}')

plt.title('Fabry-Perot cavity transmission for different finesse values')
plt.xlabel('Frequency')
plt.ylabel('Transmission')
plt.legend()

plt.show()
```
